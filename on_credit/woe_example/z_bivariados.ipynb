## **BIVARIADOS.**

Este Notebook genera el análisis bivariado tanto por quintiles como por deciles de las variables preliminares seleccionadas.

```python
from ai4b_bigquery import AI4BBigQuery
ai4b_bigquery = AI4BBigQuery('teco-prod-adam-dev-826c')
from utils import calcular_bivariados, exportar_multiples_bivariados_excel
```

```python
# Lectura de df
query = '''
--sql
SELECT *
FROM `teco-prod-adam-dev-826c.sdb_datamining.tmp_ppaycapturacreditotecopos_ajustef_train_preabt_DG`
'''

df = ai4b_bigquery.run_job_batch(script=query).to_dataframe()
df.head(1)
```

```python
df.info()
```

```python
df_biv_q = calcular_bivariados(
    df=df,
    target_col='target',
    n_bins=5,
    metodo='quantile',
    excluir_cols=['cuil_cuit', 'periodo']
    # forzar_categoricas=['tpv_tendencia_m0_vs_m_1', 'sp_tendencia_m0_vs_m_1']
)
df_biv_q.head(1)
```

```python
df_biv_d = calcular_bivariados(
    df=df,
    target_col='target',
    n_bins=10,
    metodo='quantile',
    excluir_cols=['cuil_cuit', 'periodo']
    # forzar_categoricas=['tpv_tendencia_m0_vs_m_1', 'sp_tendencia_m0_vs_m_1']
)
df_biv_d.head(1)
```
